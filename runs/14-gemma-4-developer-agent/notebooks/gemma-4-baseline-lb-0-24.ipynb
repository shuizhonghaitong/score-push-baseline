# Gemma 4 Developer Agent: v5 (safe)

Base: `flexonafft/gemma-4-budget-aware-coding-agent-v2` (public 0.22). Built after v4 hit *Notebook Timeout* / *Notebook Threw Exception*.

Config changes vs 0.22 (safety margins only):
- `max_time_minutes` 8 -> **7**: worst case 120 x 7 = 14 h instead of 16 h, more slack under the 12 h cap.
- `max_output_tokens` 8192 -> **6144**: vLLM rejects prompt + max_output > 32,768, so this leaves ~2k more tokens for history. Thinking budget (4096) still fits.
- Tools, model, temperature, 28 calls, 240 s timeout, thinking on: unchanged.

Prompt changes vs 0.22: protect `pytest.ini`/`conftest.py`, scratch files only in `/tmp`; fix every parallel implementation (sync/async, Client/AsyncClient, `__all__`); final check `git status --short && git diff`; keep current behaviour for inputs the issue does not mention; pytest output piped through `tail -40`. v4's whole-test-file run is **removed** (main suspect for the timeout).

Run on CPU, internet off. The notebook only writes `submission.zip`.
