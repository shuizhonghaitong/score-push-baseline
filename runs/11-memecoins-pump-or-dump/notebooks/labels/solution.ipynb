# Labels mirror (small)
Copies the CSV metadata/label files to /kaggle/working/labels for local model iteration.

```
import os, shutil, glob
src = None
for cand in ["/kaggle/input/competitions/memecoins-pump-or-dump", "/kaggle/input/memecoins-pump-or-dump"]:
    if os.path.isdir(cand):
        src = cand
        break
print("src:", src, flush=True)
out = "/kaggle/working/labels"
os.makedirs(out, exist_ok=True)
for f in ["train.csv", "test.csv", "train_archive.csv", "sample_submission.csv",
          "pfamm_migrations.csv", "pumpfun_token_creation.csv"]:
    p = os.path.join(src, f)
    if os.path.exists(p):
        shutil.copy2(p, os.path.join(out, f))
        print("copied", f, os.path.getsize(p), flush=True)
print("LABELS DONE", flush=True)
```
